```python
#Download audio , split in 1 minute audios, then generate a random number from 1 to 5 (nr of tts insertions)
#then save the audios in PF_malicious at the end of the loop delete the big old audio file
import librosa
import os
import random
from gtts import gTTS
import tqdm
import soundfile as sf
def process_audio(validation_path):

  scam_phrases = [
        "Suspicious login detected on your bank profile. Confirm your identity now.",
        "A payment of four hundred pounds was authorized from your visa. Press one to dispute.",
        "Your credit card has been flagged for fraudulent international transfers.",
        "To unlock your account, please enter your five-digit secret pin.",
        "Final notice: your mortgage payment is overdue. Connect to an agent.",
        "Federal agents are en route to your address. Connect to an officer immediately.",
        "Your Social Security Number has been linked to a drug trafficking investigation.",
        "Verify your identity to avoid immediate asset seizure and legal prosecution.",
        "A warrant for your arrest has been signed by the local magistrate.",
        "Tax evasion charges have been filed against you. Pay the fine to cancel.",
        "Your network firewall has been breached. System shutdown will begin shortly.",
        "Your device is infected with spyware. Remote access is required for cleanup.",
        "Critical security update failed. Your personal files are now at risk.",
        "We have detected illegal activity originating from your IP address.",
        "Microsoft support alert: your license key has expired. Call us now."
        ]

  print(f"Processing data from '{validation_path}'...")

  for category in ['PF_malicious']:
    dir_path = os.path.join(validation_path, category)

    if not os.path.exists(dir_path):
      continue
    valid_files = [f for f in os.listdir(dir_path) if f.lower().endswith(('.wav', '.mp3'))]

    print(f"\nProcessing category: {category.upper()} ({len(valid_files)} files)")

    for f in tqdm.tqdm(valid_files, desc=f"Procesare {category}"):
      old_path = os.path.join(dir_path, f)
      path=old_path
      audio1 = AudioSegment.from_wav(path)
      seconds=len(audio1)//1000

      idx=1
      audio_name=f.split(".wav")[0].strip()
      start=0
      step=60
      for start in range(0,seconds,step):
        end = min(start + step, seconds)
        audio_segment=audio1[start*1000:end*1000]

        if category=="PF_malicious":
          processed_segment = AudioSegment.empty()
          for i in range(0,len(audio_segment),6000): # I want every 6 seconds to be an insertion
            window = audio_segment[i : i + 6000]
            phrase = random.choice(scam_phrases)
            length=len(phrase)
            phrase=phrase[random.randint(0,length//2):random.randint(length//2+1,length)]

            tts = gTTS(text=phrase, lang='en')
            tts_buf = io.BytesIO()
            tts.write_to_fp(tts_buf)
            tts_buf.seek(0)

            fake_seg = AudioSegment.from_file(tts_buf, format="mp3")

            if len(window) > len(fake_seg):
              p = random.randint(0, len(window) - len(fake_seg))
              window = window[:p] + fake_seg + window[p + len(fake_seg):]
            processed_segment += window

          audio_segment = processed_segment

        buf = io.BytesIO()
        audio_segment.export(buf, format="wav")
        buf.seek(0)

        audio, _ = librosa.load(buf, sr=16000)
        save_file_name = f"{audio_name}_{idx}.wav"
        save_path = os.path.join(validation_path, category, save_file_name)

        idx+=1

        sf.write(save_path,audio,16000)
      try:
        os.remove(old_path)
      except Exception as e:
        print(f"Couldn't delete {old_path}: {e}")




process_audio("/content/drive/MyDrive/MaliciousCallDetection/dataset/validation")
```
