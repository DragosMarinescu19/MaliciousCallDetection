```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("kienngx/callhome-dataset")

print("Path to dataset files:", path)
```

```python
import os
from pydub import AudioSegment
import math

input_folder='dataset/kagglehub'
output_folder='dataset/dataset/normal'

slice_duration=5*1000 # 5 seconds
overlap=1.5*1000
step=slice_duration-overlap
if not os.path.exists(input_folder):
         raise FileNotFoundError(f"Nu gasesc folderul: {input_folder}")
    
for file_name in os.listdir(input_folder):
    if file_name.endswith('.wav'):
        file_path = os.path.join(input_folder, file_name)
        
        audio = AudioSegment.from_wav(file_path)
        duration_ms = len(audio)
                
        start = 0
        counter = 0
        
        while start + slice_duration <= duration_ms:
            end = start + slice_duration
            slice_audio = audio[start:end]
            
            output_filename = f"normalAudio5sec_{counter}.wav"
            output_path = os.path.join(output_folder, output_filename)
            
            slice_audio.export(output_path, format="wav")
            
            # step=3.5 seconds,we leave room for overlap
            start += step
            counter += 1


    
```
