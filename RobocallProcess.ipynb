```python
from datasets import Dataset, Audio
import pandas as pd
import shutil
import re

df = pd.read_csv('dataset/robocall-audio-dataset-main/robocall-audio-dataset-main/metadata.csv')

audio_dataset = Dataset.from_dict({
    "audio": df['file_name'].to_list(),
    "transcript": df['transcript'].to_list(),
    "language" : df['language'].to_list(),
    "case_pdf" : df['case_pdf'].to_list(),
}).cast_column("audio", Audio(sampling_rate=16000,decode=False))
path="dataset/robocall-audio-dataset-main/robocall-audio-dataset-main/"
dest="dataset/dataset/malicious/"
for audio in audio_dataset:
    if audio['language']=='en':
        #I need to move to "dataset/malicious" every wav file
        small_path=audio['audio']['path']
        from_path=path+small_path
        shutil.copy(from_path,dest)
```
